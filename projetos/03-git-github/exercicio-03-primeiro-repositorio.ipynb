## Parte 3 — Criando o repositório privado no GitHub

1. No GitHub, clique no `+` no canto superior direito e escolha **New repository**.
2. Dê o nome que você registrou na Parte 1, por exemplo `extracao-dados-trabalhos-seunome`.
3. Selecione **Private**.
4. Não marque nenhuma opção de criar `README`, `.gitignore` ou `license`: seu repositório local já tem tudo isso, e criar de novo no GitHub geraria conflito.
5. Clique em **Create repository** e deixe a página aberta, ela mostra a URL que você usa a seguir.

De volta ao terminal, dentro de `extracao-dados-trabalhos-SeuNome`:

```bash
git branch -M main
git remote add origin <URL_DO_REPOSITORIO>
git push -u origin main
```

Atualize a página do GitHub. Você deve ver `README.md`, `.gitignore`, a pasta `projetos/` e os 4 commits no histórico.

## Parte 4 — Me adicionando como colaborador

O repositório é **privado**: por padrão, só você o vê. Para eu conseguir corrigir sem que a turma tenha acesso às suas respostas, adicione-me como colaborador direto no repositório, e não deixe o repositório público.

1. Na página do repositório no GitHub, clique em **Settings**.
2. No menu à esquerda, clique em **Collaborators** (pode pedir sua senha de novo).
3. Clique em **Add people**.
4. Digite `mateuspestana`, selecione o usuário na lista e confirme o convite.

**OBS:** o convite fica pendente até eu aceitar, isso é normal e não depende de você. Não precisa reenviar nem se preocupar se ele não aparecer como colaborador imediatamente.

**ATENÇÃO:** nunca troque a visibilidade do repositório para **Public** para "facilitar" a correção. Isso expõe suas respostas para a turma inteira. Se isso acontecer sem querer, vá em **Settings > General > Danger Zone > Change visibility** e volte para **Private** imediatamente.
